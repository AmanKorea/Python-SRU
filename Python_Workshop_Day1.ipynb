```python
#print:write python code for energy price calculation. Units will be provided by user as input. do not use function. Only if-else statement
#example:price=units*2 for residential customers
#example:price=unit*5 for commertial customer
#example:price=unit*10 for industrial customer

units = int(input("Enter the number of units consumed: "))
customer_type = input("Enter customer type (residential, commercial, industrial): ").lower()

price = 0

if customer_type == 'residential':
  price = units * 2
  print(f"For a residential customer, the total price is: ${price}")
elif customer_type == 'commercial':
  price = units * 5
  print(f"For a commercial customer, the total price is: ${price}")
elif customer_type == 'industrial':
  price = units * 10
  print(f"For an industrial customer, the total price is: ${price}")
else:
  print("Invalid customer type entered. Please choose from residential, commercial, or industrial.")
```
