### Armstrong Number Check

```python
num_str = input("Enter a number: ") # Prompt the user to enter a number and store it as a string

# Validate input to ensure it's a number
if not num_str.isdigit(): # Check if the entered string consists only of digits
    print("Invalid input. Please enter a positive integer.") # If not, print an error message
else:
    num = int(num_str)  # Convert the valid string input to an integer
    order = len(num_str)  # Get the number of digits in the input number (its 'order')

    sum_of_powers = 0  # Initialize a variable to store the sum of the powers of digits
    temp_num = num  # Create a temporary variable to manipulate without changing the original number

    while temp_num > 0:  # Loop as long as the temporary number is greater than 0
        digit = temp_num % 10  # Get the last digit of the temporary number
        sum_of_powers += digit ** order  # Add the digit raised to the power of 'order' to the sum
        temp_num //= 10  # Remove the last digit from the temporary number (integer division)

    if num == sum_of_powers:  # Compare the original number with the calculated sum of powers
        print(f"{num} is an Armstrong number.") # If they are equal, it's an Armstrong number
    else:
        print(f"{num} is not an Armstrong number.") # Otherwise, it's not an Armstrong number
```

```python
#While_else statement
#Else statement only occures if no break conditions.
#if break statment executed, then else will not come.

List_of_Books=["python", "Java", "c", "c++"]
book=input("Enter book you want in library")
book_no=0 #start from index 0
while(book_no < len(List_of_Books)): #loop conditions corrected to avoid out-of-bounds access
  if(book==List_of_Books[book_no]):
    print("yes, book available")
    break
  book_no=book_no+1
else:
  print("no, book not available")
```
